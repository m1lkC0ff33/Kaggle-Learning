```python
import pandas as pd
import numpy as np
import sys
import os
from xgboost import XGBClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score

sys.path.append('../../../utils')
from LabelEncode import auto_label_encode
from SubmissionHelper import save_submission

train = pd.read_csv('../data/train.csv')
test = pd.read_csv('../data/test.csv')
```

```python
target_map = {'Low': 0, 'Medium': 1, 'High': 2}
train['Irrigation_Need'] = train['Irrigation_Need'].map(target_map)

train_encoded, test_encoded, _ = auto_label_encode(
    train, 
    test, 
    target_col='Irrigation_Need', 
    exclude_cols=['id']
)

X = train_encoded.drop(['id', 'Irrigation_Need'], axis=1)
y = train_encoded['Irrigation_Need']
X_test = test_encoded.drop(['id'], axis=1)
```

```python
model = XGBClassifier(n_estimators=100, random_state=42)
model.fit(X, y)

test_preds_num = model.predict(X_test)

inv_target_map = {0:'Low', 1:'Medium', 2:'High'}
final_labels = [inv_target_map[p] for p in test_preds_num]

save_submission(final_labels, test, 'id', 'Irrigation_Need', prefix='xgboost_basic')
```
